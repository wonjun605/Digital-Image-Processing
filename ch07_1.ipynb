```python
Img = cv2.imread('/content/gdrive/My Drive/ColabIP26/Images/lena256.jpg')
RGB_Img = cv2.cvtColor(Img, cv2.COLOR_BGR2RGB)

gImg= cv2.cvtColor(RGB_Img, cv2.COLOR_RGB2GRAY)
print(Img.shape)
cv2_imshow(gImg)
```

```python
height, width = gImg.shape
print(height, width)
```

```python
T = np.float32([[1,0,100],[0,1,50]])
outImg = cv2.warpAffine(gImg, T, (width, height))
cv2_imshow(outImg)
```

```python
R = cv2.getRotationMatrix2D((0,0), 45, 1)
outImg = cv2.warpAffine(gImg, R, (width, height))
cv2_imshow(outImg)
```
